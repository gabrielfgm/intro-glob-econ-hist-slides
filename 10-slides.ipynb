## How to measure inequality?


```{tikz}
#| echo: false
#| fig-align: center
#| fig-ext: svg

\begin{tikzpicture}[scale=5, font=\sffamily]
  % Axes
  \draw[->, thick] (0,0) -- (1.1,0) node[right] {Cumulative Share of Population ($x$)};
  \draw[->, thick] (0,0) -- (0,1.1) node[above] {Cumulative Share of Income ($L(x)$)};

  % Unit Square box
  \draw[dashed, gray] (0,0) rectangle (1,1);

  % Line of Perfect Equality
  \draw[thick, blue] (0,0) -- (1,1) node[above right, align=left] {Line of Perfect\\Equality ($y=x$)};

  % Lorenz Curve (using a quadratic for a smooth curve)
  \draw[thick, red] (0,0) to[out=10, in=250] (1,1);
  \node[red, below right] at (0.6,0.36) {Lorenz Curve $L(x)$};

  % Labels for 0 and 1
  \node[below] at (0,0) {0};
  \node[left] at (0,0) {0};
  \node[below] at (1,0) {1};
  \node[left] at (0,1) {1};

  % Fill Area A (between line of equality and Lorenz curve)
  \begin{scope}
    \clip (0,0) to[out=10, in=250] (1,1) -- (0,1) -- cycle;
    \fill[blue!20, opacity=0.5] (0,0) -- (1,1) -- (0,1) -- cycle;
  \end{scope}
  \begin{scope}
      \clip (0,0) to[out=10, in=250] (1,1) -- (1,0) -- cycle;
      \fill[white] (0,0) -- (1,1) -- (1,0) -- cycle;
  \end{scope}
  % A simpler way to fill A
  \fill[blue, opacity=0.3] (0,0) -- (1,1) to[out=250, in=10] (0,0);
  \node at (0.35, 0.2) {\textbf{A}};

  % Fill Area B (under Lorenz curve)
  \fill[red, opacity=0.3] (0,0) to[out=10, in=250] (1,1) -- (1,0) -- cycle;
  \node at (0.7, 0.15) {\textbf{B}};

\end{tikzpicture}
```


## The stylized facts on inequality: global income inequality  {.smaller}

::::: columns
::: {.column width="30%"}
@worldin2022

-   Inequality *globally* rises across the 19th century
-   Falls in mid-20th and rises again
-   Measured as Top-10th/Bottom-50th
:::

::: {.column width="70%"}
![@worldin2022, Figure 5](images/World-Inequality-2022-Figure-5.png)
:::
:::::

## The stylized facts on inequality: within vs between country  {.smaller}

::::: columns
::: {.column width="30%"}
@worldin2022

-   This is *not* just a story about rich vs poor *countries*
:::

::: {.column width="70%"}
![@worldin2022, Figure 6](images/World-Inequality-2022-Figure-6.png)
:::
:::::

## Global wealth growth rates: the 'elephant' curve  {.smaller}

::::: columns
::: {.column width="30%"}
@worldin2022

-   Globally seems wealthy-nation middle classes may have been squeezed
    in last 30 years by poor-nation middle classes
:::

::: {.column width="70%"}
![@worldin2022, Figure 9](images/World-Inequality-2022-Figure-9.png)
:::
:::::

## US inequality

::::: columns
::: column
Saez & Zucman (2020) *Journal of Economic Perspectives*

-   U-pattern across the 20th century based on national accounts data
:::

::: column
![:scale 85%](images/Saez-Zucman-fig-1.png)
:::
:::::

## The much-debated search for causes

::::: columns
::: column
#### Technology

-   "winner-takes-all" technologies

#### Politics

-   Rent-seeking/de-regulation/dismantling unions/dismantling anti-trust
:::

::: column
#### Globalization

-   Larger markets, bigger returns to scale?

#### Education

-   The widening skills gap? (only really works for income not wealth)
:::
:::::

#### The Piketty idea?

$r > g$ ?

## Works Cited
